# Password Strength Checker

Enter a password and let the program analyze how strong it is.

**It checks:**
- Password length
- Uppercase letters
- Lowercase letters
- Numbers
- Special characters

**It classifies:**
- Very Weak
- Weak
- Medium
- Strong
- Very Strong

## Step 1 — Get to know the string methods

Python strings have built-in methods that return `True` or `False`. They are the foundation of our analyzer.

| Method | What it checks |
|--------|----------------|
| `"A".isupper()` | Is it an uppercase letter? |
| `"a".islower()` | Is it a lowercase letter? |
| `"5".isdigit()` | Is it a digit? |

Special characters are anything that is **not** a letter or a digit. We check that with `not char.isalnum()`.

```python
samples = ["A", "z", "3", "@", " ", "!"]

print(f"{'Char':<6} {'isupper':<9} {'islower':<9} {'isdigit':<9} {'isalnum':<9} {'special'}")
print("-" * 60)

for ch in samples:
    print(
        f"{ch!r:<6} {str(ch.isupper()):<9} {str(ch.islower()):<9} "
        f"{str(ch.isdigit()):<9} {str(ch.isalnum()):<9} {str(not ch.isalnum())}"
    )
```

## Step 2 — The basic analyzer

Walk through every character in the password and use `boolean` flags to track which criteria are met.

The final strength depends on how many criteria pass:

```python
def strength_from_checks(checks):
    """Map the number of passed checks (0–5) to a label."""
    labels = {0: "VERY WEAK", 1: "VERY WEAK", 2: "WEAK",
              3: "MEDIUM", 4: "STRONG", 5: "VERY STRONG"}
    return labels[checks]
```

```python
def analyze_basic(password):
    """Check each criterion with a for-loop and boolean flags."""
    length = len(password)

    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False

    for char in password:
        if char.isupper():
            has_upper = True
        elif char.islower():
            has_lower = True
        elif char.isdigit():
            has_digit = True
        else:
            has_special = True

    checks = sum([length >= 8, has_upper, has_lower, has_digit, has_special])

    print("========== PASSWORD ANALYZER ==========\n")
    print(f"Password Length       : {length}")
    print(f"Uppercase Letters     : {'\u2713' if has_upper else '\u2717'}")
    print(f"Lowercase Letters     : {'\u2713' if has_lower else '\u2717'}")
    print(f"Numbers               : {'\u2713' if has_digit else '\u2717'}")
    print(f"Special Characters    : {'\u2713' if has_special else '\u2717'}")
    print(f"\nStrength              : {strength_from_checks(checks)}")
    print("\n" + "=" * 40)
```

```python
password = input("Enter password: ")
analyze_basic(password)
```

## Step 3 — Optional challenge: password score

Instead of a simple pass/fail per rule, assign a **point value** to each criterion and sum them up.

| Rule | Points |
|------|--------|
| Length >= 12 | 3 |
| Length >= 8 | 2 |
| Length >= 5 | 1 |
| Uppercase letters | 2 |
| Lowercase letters | 2 |
| Numbers | 2 |
| Special characters | 2 |

**Score table:**

| Score | Strength |
|-------|----------|
| 0–2 | Very Weak |
| 3–4 | Weak |
| 5–6 | Medium |
| 7–8 | Strong |
| 9+ | Very Strong |

```python
def password_score(password):
    """Return a score out of 11 for the given password."""
    score = 0

    # --- length ---
    length = len(password)
    if length >= 12:
        score += 3
    elif length >= 8:
        score += 2
    elif length >= 5:
        score += 1

    # --- character types ---
    has_upper    = any(c.isupper() for c in password)
    has_lower    = any(c.islower() for c in password)
    has_digit    = any(c.isdigit() for c in password)
    has_special  = any(not c.isalnum() for c in password)

    if has_upper:   score += 2
    if has_lower:   score += 2
    if has_digit:   score += 2
    if has_special: score += 2

    return score
```

```python
def strength_from_score(score):
    """Map a score (0–11) to a strength label."""
    if score <= 2:
        return "VERY WEAK"
    elif score <= 4:
        return "WEAK"
    elif score <= 6:
        return "MEDIUM"
    elif score <= 8:
        return "STRONG"
    else:
        return "VERY STRONG"
```

```python
def analyze_scored(password):
    """Full analysis using the scoring system."""
    length = len(password)
    has_upper   = any(c.isupper() for c in password)
    has_lower   = any(c.islower() for c in password)
    has_digit   = any(c.isdigit() for c in password)
    has_special = any(not c.isalnum() for c in password)

    score    = password_score(password)
    strength = strength_from_score(score)

    print("========== PASSWORD ANALYZER ==========\n")
    print(f"Password Length       : {length}")
    print(f"Uppercase Letters     : {'\u2713' if has_upper else '\u2717'}")
    print(f"Lowercase Letters     : {'\u2713' if has_lower else '\u2717'}")
    print(f"Numbers               : {'\u2713' if has_digit else '\u2717'}")
    print(f"Special Characters    : {'\u2713' if has_special else '\u2717'}")
    print(f"\nScore                 : {score}/11")
    print(f"Strength              : {strength}")
    print("\n" + "=" * 40)
```

## Step 4 — Test it

Enter your own password below to see the scored analysis.

```python
password = input("Enter password: ")
analyze_scored(password)
```
